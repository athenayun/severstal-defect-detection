# Severstal Steel Defect Detection

## 1. Project Overview

This project aims to develop a supervised computer vision
system for automated steel surface inspection.

The system identifies defective steel surfaces, predicts
defect categories, and localizes defective regions using
pixel-level segmentation.

## 2. Machine Learning Objectives

- Binary classification: normal vs. defective steel surfaces.
- Defect segmentation: identify the location and category
  of each defect at the pixel level.

## 3. Modeling Strategy

- Baseline: ResNet18 with transfer learning.
- Main model: supervised segmentation network.
- Framework: PyTorch.
- Computing environments: Apple MPS and Kaggle GPU.
